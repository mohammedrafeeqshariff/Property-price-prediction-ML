```python
import pandas as pd
ds = pd.read_csv("../data/melb_data.csv")
```

```python
#to look howmany null datas.
print(pd.isnull(ds).sum())
```

To view null values visually.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(ds.isnull(), cbar=False, cmap="YlOrRd")
plt.title("Missing Value Heatmap")
plt.show()
```

To fill not available rows into a much relative value

```python
ds[BuildingArea].fillna(ds[BuildingArea].mean(), inplace=True)
ds[YearBuilt].fillna(ds[YearBuilt].mean(), inplace=True)
ds[Car].fillna(ds[Car].mean(), inplace=True)
ds[CounsilArea].fillna(ds[CounsilArea].mode()[0], inplace=True)
```
